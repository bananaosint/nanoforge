# NanoForge Studio — Kaggle T4 x2

Select **GPU T4 x2** and turn **Internet on** in Kaggle settings, then run this cell. It verifies both GPUs and launches the full UI. Model size, datasets and training settings stay in the UI; nothing trains automatically.

Open the printed link in a new tab. Username is `nanoforge`; the random password is printed below it. The temporary Cloudflare tunnel requires that password for every page/API. Keep notebook output private. The first hardware tests exercise both GPUs before opening Studio.

Each GPU holds a full model replica; their memory is not pooled. Choose at least batch 2 when **Use both GPUs** is enabled. Save/download checkpoints before the Kaggle session ends. Use `kaggle_studio.stop()` to close the tunnel and safely stop the trainer.


```python
from pathlib import Path
import subprocess
from IPython.display import HTML, display

if 'kaggle_studio' in globals():
    kaggle_studio.stop()

REPO = Path('/kaggle/working/nanoforge')
if not REPO.exists():
    subprocess.run(['git','clone','https://github.com/bananaosint/nanoforge.git',str(REPO)],check=True)
else:
    subprocess.run(['git','-C',str(REPO),'pull','--ff-only'],check=True)

startup = {'__name__': 'nanoforge_kaggle_ui'}
script = REPO/'scripts/kaggle_ui.py'
exec(compile(script.read_text(),str(script),'exec'),startup)
kaggle_studio = startup['launch_kaggle'](repo=REPO)
display(HTML(f'<a href="{kaggle_studio.url}" target="_blank" rel="noopener">Open NanoForge Studio</a>'))
```
